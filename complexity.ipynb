n time: linear

```python
def sum_to(n):
    total = 0
    for i in range(1, n +1 ):
        total += i
    return total

# run n time
```

```python
sum_to(100)
```

```python
sum_to(4)
```

n * n time: quadratic(n^2)

```python
def square_count(n):
    count = 0
    for i in range(n):
        for j in range(n):
            count += 1
    return count

# run n * n time
```

```python
square_count(3)
```

```python
square_count(4000)
```

pairs:

```python
def pairs_loop(n):
    count = 0
    for i in range(n):
        for j in range(i + 1, n):
            count += 1
    return count
```

```python
pairs_loop(10)
```

largest n for quadratic: n^2

```python
def largest_n_quadratic(budget):
    n = 0
    while (n + 1) * (n + 1) <= budget:
        n += 1
    return n
```

```python
largest_n_quadratic(100)
```

halving gives a log: O(log n)

```python
def halving_count(n):
    count = 0
    while n > 1:
        n //= 2
        count += 1
    return count
```

```python
halving_count(100)
```

doubling: O(log n)

```python
def doubling_count(n):
    count = 0
    i = 1
    while i < n:
        i *= 2
        count +=1
    return count
```

```python
doubling_count(100)
```

```python
def is_prime(n):
    if n < 2:
        return False
    i = 2
    while i * i <= n:
        if n % i == 0:
            return False
        i += 1
    return True
```

```python
is_prime(97)
```

```python
def isqrt_search(n):
    lo, hi = 0, n
    while lo < hi:
        mid = (lo + hi + 1)//2
        if mid * mid <= n:
            lo = mid
        else:
            hi = mid - 1
    return lo
```

```python
isqrt_search(6)
```

```python
isqrt_search(16)
```
