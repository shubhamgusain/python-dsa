```python
def sum_to(n):
    total = 0
    for i in range(1, n +1 ):
        total += i
    return total
```

```python
sum_to(100)
```

```python
sum_to(4)
```

```python
def square_count(n):
    count = 0
    for i in range(n):
        for j in range(n):
            count += 1
    return count
```

```python
square_count(3)
```

```python
square_count(4000)
```
